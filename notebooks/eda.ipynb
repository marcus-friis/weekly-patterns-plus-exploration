# Exploratory data analysis of [*Weekly Patterns Plus*](https://app.deweydata.io/data/advan/weekly-patterns-plus)
From *Advan Research*
> Weekly foot traffic to points of interest across the United States and Canada, with each record covering a single POI over one Monday-to-Sunday week and reporting aggregated visit and visitor counts alongside visitor home-location and demographic aggregations. This dataset also ships a crosswalk to Advan's legacy Patterns store IDs for linking across Advan's Patterns generations.

**Observation Level:**
POI x week - one row per point of interest per Monday-to-Sunday week.

```python
import itertools

import geopandas as gpd
import matplotlib.pyplot as plt
import polars as pl
import seaborn as sns
from shapely.geometry import Point

from advan_test import plot
from advan_test.data import load_weekly_patterns_plus
from advan_test.utils import project_root
```

```python
ROOT = project_root()
DATA_PATH = ROOT / "data"
CSV_PATH = DATA_PATH / "weekly-patterns-plus-sample.csv"
US_PATH = DATA_PATH / "us-states.json"
```

```python
df = load_weekly_patterns_plus(CSV_PATH)
df.head()
```

```python
states = gpd.read_file(US_PATH)
points = gpd.GeoDataFrame(
    df.to_pandas(),
    geometry=[Point(xy) for xy in zip(df["longitude"], df["latitude"])],
    crs="EPSG:4326",
)
```

```python
dimensions = ["city", "region", "iso_country_code"]
n = len(dimensions)
fig, axes = plt.subplots(1, n)
for i in range(n):
    dim = dimensions[i]
    ax = axes[i]
    plot.plot_pois_by_dimension(df, dim, 10, ax=ax)
fig.tight_layout()
```

```python
# mean of median is kind of weird
agg = df.group_by("id_store").agg(
    pl.col("visitor_counts").mean().alias("mean_weekly_visitors"),
    pl.col("median_dwell").mean().alias("mean_median_dwell"),
)
sns.scatterplot(data=agg, x="mean_weekly_visitors", y="mean_median_dwell")
```

```python
regions = df["region"].unique()
n = len(regions)
num_cols = 5
num_rows = -(-n // num_cols)  # ceil division
states_joined = states.sjoin(points)

fig, axes = plt.subplots(num_rows, num_cols, figsize=(24, 18))
axes_flat = axes.flatten()

for ax, region in itertools.zip_longest(axes_flat, regions):
    if region is None:
        ax.axis("off")  # hide unused subplot
        continue
    points_filtered = points.loc[points["region"] == region]
    states_filtered = states_joined.loc[states_joined["region"] == region]
    ax.set_title(region)
    plot.plot_poi_locations(points_filtered, states_filtered, ax=ax)

fig.tight_layout()
```
